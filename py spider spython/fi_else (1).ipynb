```python
# check user Enter character is Spical Character or not
char=input("Enter a Alphabet :")
if "A"<= char<='Z' or 'a'<=char <='z' or '0'<=char<='9':
    print(f"{char} is a not Spical Character ")
else:
     print(f"{char} is a Spical Character")
    
#0r 
if char.isalnum():
    print(f"{char} is a not Spical Character ")
else:
     print(f"{char} is a Spical Character")
```

```python
#check Given Data  is Mutable or imutable 
data=eval(input("Enter Any Type of Value :"))
if type(data) in (list, str, dict, set):
    print(f"{data} is Mutable")
else:
    print(f"{data} is IMutable")
```

```python
# check given word is key word or not
import keyword as kw
word=input("Enter a word")
if word in kw.kwlist:
    print(f'{word} is a Key word')
else:
     print(f'{word} is not a Key word')
```
