```python
class SBI:
    name ='state bank of india'
    location="BTM Layout"
    manager="Dineshkumar"
    ifsc_code='SBIN0002806'

    def __init__(self,name,account_no,pin,balance):
        self.name=name
        self._account_no=account_no
        self.__pin=pin
        self.transactions=[]
        self.balance=balance

    def deposits(self,amount,name):
        print(f"depositing Rs {amount} into {name} Account")
        self.balance+=amount
        self.transactions.append(f"deposited Rs {amount}")

    def withdraw(self,pin,amount):
        if self.__pin==pin:
            if self.balance>amount:
                print(f" {amount} Amount Debited")
                self.balance-=amount
                self.transactions.append(f"withdraw Rs {amount}")
            else:
                print('insufficent balane')
        else:
            print("wrong pin")


    def display_teansactions(self):
        for transaction in self.transactions:
            print(transaction)
obj=SBI('Dinesh',987654321,1234,5000)
obj.deposits(1000,'Dinesh')
obj.withdraw(1234,1000)
obj.display_teansactions()
    
```
