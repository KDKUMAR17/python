```python
def int_type_check(func_name):
    def wrapper(*args,**kwargs):
        int_type=True
        for val in args:
            if type(val)!=int:
                int_type=False
                break
        else:
             for val in kwargs.values():
                if type(val)!=int:
                    int_type=False
                    break
        if int_type:
            return func_name(*args,**kwargs)
        else:
            return "invalid values cannot execute"
    return wrapper

@ int_type_check
def add(n1:int,n2:int):
    return n1/n2

print(add(7,9))
```

```python
from time import time
def get_execution_time(func_name):
    def wrapper(*args,**kwrgs):
        print("in wrapper")
        start=time()
        result=func_name(*args)
        end=time()
        comp=end-start
        return comp
    return wrapper
@get_execution_time
def real(start ,end,step=1):
    list=[]
    for val in range(start,end,step):
        list.append(val)
    return list
real(1,900000)

```

```python

def class_deco(cls_add):
    def decorater(func_add):
        def wrapper(*args,**kwargs):
            print("*"*10,"\n")
            print(func_add(*args,**kwargs))
            print("*"*10,"\n")
        return wrapper
    for name,add in cls_add.__dict__.items():
        if(callable(add)):
            setattr(cls_add,name,decorater(add))
    return cls_add
@class_deco
class Math:
    def __init__ (self,num1,num2):
        self.num1=num1
        self.num2=num2
        return "in init"
    def add(self):
        return self.num1+self.num2
    def sub(self):
        return self.num1-self.num2
    def mul(self):
        return self.num1*self.num2

obj=Math(10,20)
obj.add()
obj.sub()
obj.mul()
```
