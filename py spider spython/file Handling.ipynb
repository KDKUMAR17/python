```python
def filter_files(file_name,message,newfile):
    file=open(file_name)
    list1=[]
    for line in file:
       if line.strip()!= "":
           s=line.split()
           if len(s)>2:


            msg=s[2]
        
            if msg == message:
                list1.append(line)

    file.close()
    file=open(newfile,'a')
    file.writelines(list1)
    file.close()
    
    
filter_files('simple.log','INFO','event.log')

```

```python
def sep_file():
    file=open('simple.log')
    info=[]
    event=[]
    trace=[]
    warning=[]
    unknown=[]

    for line in file:
         if line.strip()!='':
              
              msg=line.split()
              if len(msg)>2:
                   
                   
                   check=msg[2]
                   
                   if check =='INFO' :
                        info.append(line)
                   elif check =='EVENT' :
                        event.append(line)
                   elif check =='WARNING' :
                        
                        warning.append(line)
                   elif check =='TRACE' :
                        
                        trace.append(line)
                   else:
                        
                        unknown.append(line)
    out={'TRACE':trace,'EVENT':event,'WARNING':warning,'INFO':info,'UNKNOWN':unknown}
    
   
    file.close()
    return out
```

```python
list_type=sep_file()
for item in list_type:
    print(item)
    file=open(item +'.log','a')
    print(list_type[item])
    writ=list_type[item]
    file.writelines(writ)
    file.close()


```
