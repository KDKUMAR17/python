```python
class Aromstrong:
    def __init__(self):
        self.number=1
        
    def __iter__(self):
        return self
    def __next__(self):
        while True:
            sum=0
            power=len(str(self.number))
            temp=self.number
            while temp!=0:
                last_num=temp%10
                sum+=last_num **power
                temp//=10
            if self.number==sum:
                self.number+=1
                return sum
            self.number+=1
        
obj=Aromstrong()

    
for i in range(11):
   print(next(obj)) 
    


                





            




```

```python
def prime():
    number=2
    while True:
        for val in range(2,number):
            if number%val==0:
                break
        else:
            yield number
        number+=1
obj=prime()
num=int(input("enter NO.OF prime number you want: "))
for _ in range (num):
    print(next(obj))
        
```
