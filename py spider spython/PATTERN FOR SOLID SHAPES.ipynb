# PATTERN FOR SOLID SHAPES

```python

num=int(input("enter the number"))
for row in range(num):
    for col in range(num):
        if col==0 or col==row or row==num-1 or row>col:
            print("*",end=' ')
        else:
            print(' ',end=' ')
    print()

```
