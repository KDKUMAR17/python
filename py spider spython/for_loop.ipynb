```python
#give dict swap the key as a value, value as a key
dict={'a':10,"b":20,'c':30}
output={}
for key in dict:
    item=dict[key]
    if type(item) not in [list,dict,set]:
        output[item]=key
print(output)
```

```python
#filter a set to have a single value data type
items={15,10.5,7+2j,'hello',(10,20)}
for item in items:
    if type(item) in (set,tuple):
        items.remove(item)
print(items)
```

```python
line=input("enter a sentences")
dict={}
words=line.split()
for word in words:
    if word in dict:
        dict[word]+=1
    else:
        dict[word]=1
max_value=max(dict.values())
max_word=[]
for key in dict:
    if dict[key]==max_value:
        max_word +=[key]
print(max_word)
        
```

```python
#check the given number is prime or not
num=int(input("enter the number"))
prime=True
if num==0 or num==1:
    print(num,"not a prime")
else:
    for i in range(2,num):
        if num%i==0:
            prime=False
            break
    if prime:
        print("prime")
    else:
        print("not prime")
```

```python
#fibonaci serise
num=int(input("enter the number"))
first=-1
last=1

for _ in range(num):
    fib=first +last
    first=last
    last=fib
print(fib)
```

```python
#check the collection is homoguines or not
collection=eval(input("Enter the collection"))
ref_type=type(collection[0])
for item in collection:
    if ref_type!=type(item):
        print("not homogines")
        break
else:
    print("homogines")
```

```python
def isPrime(number):
    if(number<=1):
        return False
    for i in range(2,number):
        if (number%i==0):
            return False
    return True
num=int(input("enter The number"))
isPrime(num)
```
