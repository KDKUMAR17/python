```python
def len_of_word(*words):
    print(words)
    out = {}
    for num in range(len(words)):  # Corrected range
        if words[num] not in out:  # Corrected key check
            out[words[num]] = len(words[num])  # Store word and its length
    return out

s = len_of_word("dinesh", 'kumar', 'ram')  
print(s)
```

```python
def len_word(*words):
    out={}
    for word in words:
        if word not in out :
            out[word]=len(word)

    return out
print(len_word('thala','dinesh','thala','kumar'))
```

```python
def str_pop(**kwargs):
    lists=[]
    for key in kwargs:
        if type(kwargs[key])==str:
            lists.append(key)
    for val in lists:
        kwargs.pop(val)
    print(kwargs )
str_pop(nam="dinesh",n=78,yu="raw")
```

```python
def count_up(start:int,stop:int):
    if start>stop:
        return None
    else:
        print(start)
        count_up(start+1,stop)
count_up(0,10)
```

```python
def count_down(start:int,stop:int=0):
    if start<stop:return None
    else:
        print(start)
        count_down(start -1)
count_down(10)
```

```python
def fibo(num:int):
    if num<=0  or num==1 :
        return 0
    elif num==2: 
        return 1
    else:
        
        return fibo(num-1)+ fibo(num-2)
fibo(3)
```
