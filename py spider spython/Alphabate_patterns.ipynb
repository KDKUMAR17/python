```python
num = int(input("Enter The Number: "))
char = "A"

for row in range(num):
    new = char
    for col in range(num):
        if col <= num // 2 + 1 :
            # Print increasing part
            print(new, end=' ')
            new = chr(ord(new) + 1)
        elif col >= num // 2 -1:
            # Print decreasing part
            new = chr(ord(new) - 1)
            print(new, end=' ')
    print()
    # Move to the next starting letter for the next row
    char = chr(ord(char) + 1)
```

```python
num = int(input("Enter The Number: "))
char = "A"

for row in range(num):
    new = char
    for col in range(num):
        if col >= num // 2:
            # Print increasing part
            print(new, end=' ')
            new = chr(ord(new) - 1)
        else:
            # Print decreasing part
            new = chr(ord(new) + 1)
            print(new, end=' ')
    print()
    # Move to the next starting letter for the next row
    char = chr(ord(char) + 1)
```

```python
num = int(input("Enter the Number: "))
for row in range(num):
    # Set the starting character for each row
    char = chr(ord('A') + row)
    
    # Print the increasing part
    for col in range(num // 2 + 1):
        print(char, end=' ')
        char = chr(ord(char) + 1)
    
    # Decrease character for the decreasing part
    char = chr(ord(char) - 2)
    
    # Print the decreasing part
    for col in range(num // 2):
        print(char, end=' ')
        char = chr(ord(char) - 1)
    
    # Move to the next row
    print()
```
