```python
class Prime:
    def __init__(self, end):
        self.start = 0
        self.end = end
        self.current = 2  # Start checking for primes from 2

    def __iter__(self):
        return self

    def isPrime(self, val):
        prime = True
        for num in range(2, val):
            if val % num == 0:
                prime = False
                break
        return prime

    def __next__(self):
        while self.current < self.end:
            if self.isPrime(self.current):
                prime = self.current
                self.current += 1
                return prime
            self.current += 1
        raise StopIteration

obj = Prime(100)

print(next(obj))
print(next(obj))
print(tuple(obj))



    
```

```python
def recursive_sublists(a):
    if len(a) == 0:
        return [[]]
    smaller_sublists = recursive_sublists(a[1:])
    return smaller_sublists + [[a[0]] + sublist for sublist in smaller_sublists]

a = [1, 2, 3,4,5]
print(recursive_sublists(a))
```
