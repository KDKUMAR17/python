```python
def fact():
    n1=0
    n2=1
    while True:
        yield n1
        n1,n2=n2,n1+n2
       
    
f=fact()
for i in range(5):
    print(next(f))

    
```

```python
def genprime():
    prime=2
    while True:
        for num in range(2,prime):
            if prime%num==0:
                break
        else:
            yield prime
        prime+=1
        
gen=genprime()

for i in range(10):
    print(next(gen))
```

```python
def recfib (n):
    if n==0 or n==1:
        return 1
    else:
        return n*recfib(n-1)

print(recfib(5))
```

```python
def countdown(st,ed):
    if st>ed:
        return None
    else:
        print(ed)
        return countdown(st,ed-1)
countdown(1,10)
```

```python
def gen_strong():
    temp=1
    while True:
        num=temp
        sum=0
        while num!=0:
            last=num%10
            fact=1
            
            for i in range(2,last+1):
                fact*=i
    
            sum+=fact
            num=num//10
        if temp==sum:
            yield sum
        temp+=1
s=gen_strong()
for i in range(4):
    print(next(s))
```

```python
def isprime(num,start=2,prime=True):
    if num==1:
        return False
    elif num==start:
        return prime
    else:
       if num%start==0:
           prime=False
       return isprime(num,start+1,prime)
print(isprime(4))
```

```python
def mtable(table,count,start=1):
    if count<start:
        return None
    else:
        print(f'{start} X {table} = { start*table}')
        return mtable(table,count,start+1)
mtable(10,10)
```
