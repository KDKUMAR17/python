# PATTERN FOR ALL ALPHABETS


```python
#pattern for H
num=int(input("Enter the odd Number for H pattern"))
for row in range(num):
    for col in range(num):
        if col==0 or col==num-1 or row==num//2:
            print('*',end=' ')
        else:
            print(' ',end=' ')
    print()
```

```python
# pattern for N
num=int(input("Enter the odd Number for N pattern"))
for row in range(num):
    for col in range(num):
        if row==col or col==0 or col==num-1:
            print("*",end=' ')
        else:
            print(" ",end=' ')
    print()
            
```

```python
#pattern for M
num=int(input("Enter the odd number for M pattern"))
for row in range(num):
    for col in range(num):
        if col==0 or col==num-1 or (col==row and row<=num//2)or (col+row==num-1 and row<=num//2):
            print("*",end=' ')
        else:
            print(' ',end=' ')
    print()
```

```python
# pattern for G
num=int(input('Enter the odd Number for G pattern'))
for row in range (num):
    for col in range(num):
        if row==0 or col==0 or row==num-1 or (col==num-1 and row>=num//2) or (row==num//2 and col>=num//2):
            print("*" ,end=' ')
        else:
            print(' ',end=' ')
    print()
```

```python
#pattern for W
num=int(input("Enter the odd number for W pattern"))
for row in range(num):
    for col in range(num):
        if col==0 or col==num-1 or (col==row and row>=num//2)or (col+row==num-1 and row>=num//2):
            print("*",end=' ')
        else:
            print(' ',end=' ')
    print()
```

```python
#pattern for V
num=int(input("enter the Odd number for V pattern"))
for row in range (num):
    for col in range(num):
        if (col+row==num-1 and row<=num//2) or(col==row and row<=num//2):
            print('*',end=' ')
        else:
            print(" ", end=" ")
    print()
```

```python
#pattern for Q
num=int(input ("enter the Odd number for Q pattern"))
for row in range(num):
    for col in range(num):
        if row==0 or(col==0 and row<=num//2+2) or (col==num-1 and row<=num//2+2)or row==num//2+2 or (row>num//2 and row==col) :
            print("*",end=" ")
        else:
            print(' ',end=' ')
    print()
```
