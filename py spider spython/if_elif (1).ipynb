```python
set1=eval(input("Enter the set1"))
set2=eval(input("Enter the set2"))
if set1<set2:
    print(f'{set1} is subset of {set2}')
elif set2<set1:
    print(f'{set2} is subset of {set1} ')
elif set1==set2:
    print("Two set is equal")
elif set1.isdisjoint(set2):
    print("Disjoin set")
else:
    print("No relation")
```

```python
#Greatest among three numbers
num1=int(input("Enter NO 1"))
num2=int(input("Enter NO 2"))
num3=int(input("Enter NO 3"))
if num1>num2 and num1>num3:
    print(f"{num1}  Is greatest ")
elif num2>num1 and num2>num3:
    print(f"{num2}  Is greatest ")
elif num1==num2 or num2==num3 or num3==num1:
    print("All are equal" )
else:
     print(f"{num3}  Is greatest ")
    
```
