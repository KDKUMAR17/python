```python
def aromatrong():
    number=150
    while True:
        temp=number
        leng=len(str(number))
        sum=0
        while temp!=0:
            last=temp%10
            sum+=last**leng
            temp//=10
        if sum==number:
            yield sum
        number+=1
obj=aromatrong()
for i in range(10):
    print(next(obj))
            


```

```python
def fibo(num):
    if num==1:
        return 0
    elif num in (2,3):
        return 1
    else:
        return fibo(num-1)+ fibo(num-2) 
print(fibo(5))
```

```python
def fact(num):

    fact=1
    while True:
        yield num*fact
        fact+=1
fa=fact(7)
for i in range(5):
    print(next(fa))
```

```python
def fersqur():
    num=1
    while True:
        for i in range(1,num+1):
            if num/i==i:
                yield num
                break
        num+=1
ob=fersqur()
for i in range(5):
    print(next(ob))
```
