```python
#recursive
def fact(n):
    if(n==0):
        return 1
    return n*fact(n-1)
n=int(input("NUmber :"))
print(fact(n))
```

```python
# fibonasi
n=int(input("Number :"))
a=1
b=-1
c=0
for i in range(n):
    c=a+b
    b=a
    a=c
    print(c)
```

```python
#recursive
def fib(n):
    if(n<=1):
        return n
    return fib(n-1)+fib(n-2)
n=int(input("Number :"))
print(fib(n))
```
