Q10. Write a program to check whether a number is prime or not using while loop.


```python
num=int(input("Enter the number:"))
if num>1:
    prime=True
    for divmod in range(2,num):
        if num%divmod==0:
            prime=False
            break
    if prime:
        print(f"{num} is a prime number")
    else:
        print(f"{num} is not a prime number")   

else:   
    print(f"{num} is not a prime number")
```

Q15. Write a program to print the Fibonacci series till n terms (Accept n from user) using while
loop.


```python
num=int(input("Enter the number:")) 
index=0
first=-1
second=1
while index<num:
    sum=first+second
    first=second
    second=sum
    print(sum)
    index+=1
```

Q17. Write a program to check whether a number is Armstrong or not. (Armstrong number is a
number that is equal to the sum of cubes of its digits for example : 153 = 1^3 + 5^3 + 3^3.)

```python
num=int(input("Enter the number:"))
copy=num
sum=0
while num!=0:
    last=num%10
    sum+=last**len(str(copy))
    num//=10
if sum==copy:
    print(f'{copy} is Armstrong Number')
else:
    print(f'{copy} is not Armstrong Number')
    

```

Q18. Write a program to add first n terms of the following series using a while loop:
1/1! + 1/2! + 1/3! + …….. + 1/n!


```python
num=int(input("Enter The Number"))
index=1
sum=0.0
while index<=num:
    factorial=1
    index1=index
    while index1>0:
        factorial*=index1
        index1-=1
    sum+=1/factorial
    index+=1
print(sum)
```

Q20. Write a program to enter the numbers till the user enter ZERO and at the end it should
display the count of positive and negative numbers entered.

```python
positive=0
negative=0
while True:
    num=int(input("Enter the number:"))
    if num>0:
        positive+=1
    elif num<0:
        negative+=1
    else:
        break
print(f"Number of positive numbers:{positive}")
print(f"Number of negative numbers:{negative}")
```
