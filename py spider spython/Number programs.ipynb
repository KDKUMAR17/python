```python
def isaromstrong(num):
    sum=0
    temp=str(num)
    for digit in range(len(temp)):
        print(temp[digit])
        sum+=int(temp[digit])**len(temp)
        if sum==num:
            return "Aromstromg"
        else:
            print(sum,num)
            return "not Aromstrong"
num=int(input("enter the number"))
print(isaromstrong(num))
```

```python
sum=0
num=int(input("enter the number"))
temp=str(num)
for digit in temp:
    sum+=int(digit)**len(temp)
if sum==num:
    print("Aromstromg")
else:
    print("not Aromstrong")
```

```python
num=int(input("enter the number"))
temp=num
sum=0
while num!=0:
    last_digit=num%10
    sum+=last_digit**len(str(temp))
    num//=10
if sum==temp:
    print("Aromstromg")
else:
    print("not Aromstrong")
```

```python
#strong number
from math import factorial
num=int(input("enter the number"))
temp=num
sum=0
while num!=0:
    last_digit=num%10
    sum+=factorial(last_digit)
    num//=10
if sum==temp:
    print("strong")
else:
    print("not strong")
```
