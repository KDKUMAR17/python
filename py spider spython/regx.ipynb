```python
import re
```

```python
'''? -> 0 or 1 occurences'''

string = "color colour which color"
pattern = "colou?r"
res = re.findall(pattern, string)
print(res)
```

```python
string= 'i am god,gd'
pattern='go?d'
print(re.findall(pattern,string))
```

```python
'''. -> any character (but one only)'''

string = "grey may be gray but not gruy gray is combo of black and white"
pattern = "gr.y"
print(re.findall(pattern, string))
```

```python
str='i am good'
pattern='go.d'
print(re.findall(pattern,str))
```

```python
string = "grey may be gray but not gruy gray is combo of black and white"
pattern = "gr[ea]y"
print(re.findall(pattern, string))

'''[character] -> character set(one char only)'''
string = "Smith once used to be a blacksmith not Steve Smith"
pattern = "[Ss]mith"
print(re.findall(pattern, string))
```

```python
str="Dinesh ganesh"
pat='[Dd]inesh'
print(re.findall(pat,str))
```

```python
'''+ -> meta character(one or more occurences)'''
string = 'he looked at hanna and yelled, "Hannnna!, what have you done hann nna?"'
pattern = "[Hh]ann+a"
print(re.findall(pattern, string))
```

```python
str='dinesh Dineshhhhh'
pat='[Dd]inesh+'
print(re.findall(pat,str))
```

```python
r'''\d -> char set [0123456789] one digit only'''
string = "99 people called 911 in last 20 days"
pattern = r"\d+"
# pattern = "[0123456789]+"
print(re.findall(pattern, string))
```

```python
str='my NO 12378956409'
# pat=r'\d+'
pat='[0-9]+'
print(re.findall(pat,str))
```

```python
'''[a-zA-Z] -> char set (one alphabet ony)'''
string = "XAEA-12"
pattern = "[a-zA-Z]"
print(re.findall(pattern, string))
```

```python
str='Dinesh-17'
pat='[A-Za-z]'
print(re.findall(pat,str))
```

```python
'''^ -> in char set represent negation meaning anything other than the characters mentioned in the char set'''
string = "humpty dumpty... sat on a wall"
pattern = r"[^aeiouAEIOU\W]"
print(re.findall(pattern, string))
```

```python
'''
^ -> outside char set represent startswith
$ -> represent endswith
'''

string = "mary had a little lamb his fleece was white as snow"
pattern = "^mary" #starts with mary
print(re.findall(pattern, string))
pattern = "snow$"   #Ends with snow
print(re.findall(pattern, string))
```

```python
string = "100 Rs"
pattern = r'\d+'
print(re.findall(pattern, string))
```

```python
string = "yo you young boy yelling is bad"
pattern = 'y[aeiouAEIOU][a-zA-Z]*'
print(re.findall(pattern, string))
```

```python
'''
^ and $ if used on a same pattern, checks if the pattern is a single word in that string
'''
string = "twinkle"
pattern = "^twinkle$"
print(re.findall(pattern, string))
```

```python
string = "python is king java is ... python is better than java"
pattern = "(python|java)"
print(re.findall(pattern, string))
```

```python
string = "I 210c y12 420 5we1 840 god 1680a"
# pattern = "[^a-zA-Z0-9]\d\d\d[^a-zA-Z0-9]"
pattern = r"\b\d\d\d\b"
print(re.findall(pattern, string))
```

```python
string = "Sh@zzy_is @maz!ng"
pattern = r"\b[a-zA-Z]+\b"
# pattern = r"\b\w+\b"
print(re.findall(pattern, string))
```

```python
string = "1 machli jal ki raani hai"
pattern = r"\b[a-zA-Z]+\b"
# pattern = r"\b\w+\b"
print(re.findall(pattern, string))
```

```python
string = "10 books 1 cost rs. 100 each if 20000 people buy 125 books"
pattern = r"\b\d{2,5}\b"
print(re.findall(pattern, string))
```

```python
str='127.0.0.1:8080'
pattern='(:?(:?[1-9][0-9][0-9].|[1-9][0-9].|[0-9].).{3})'
print(re.findall(pattern,str))
```
